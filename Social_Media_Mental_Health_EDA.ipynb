```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

sns.set_theme(style="darkgrid")
plt.rcParams["figure.figsize"] = (10, 6)

print("✅ Analytics and visualization libraries successfully initialized!")
```

```python
import kagglehub
import os
# Download latest version
path = kagglehub.dataset_download("shivasingh4945/student-social-media-and-mental-health-impact")


csv_file_name = os.listdir(path)[0]
csv_path = os.path.join(path, csv_file_name)


df = pd.read_csv(csv_path)
```

```python
print(df.shape)
print(df.info())
print(df.head(3))
```

```python
fig, ax = plt.subplots(figsize=(10, 6))

ax.scatter(
    x = df['Avg_Daily_Usage_Hours'],
    y = df["Sleep_Hours_Per_Night"],
    alpha = 0.5,
    color = "royalblue",
    edgecolor = "white",
)

ax.set_title("The Digital Drain: Screen Time vs. Nightly Sleep", fontsize=14, fontweight='bold', pad=15)
ax.set_xlabel("Average Daily Phone Usage (Hours)", fontsize=11)
ax.set_ylabel("Sleep Duration Per Night (Hours)", fontsize=11)

ax.grid(True, linestyle="--", alpha=0.6)

plt.show()
```
