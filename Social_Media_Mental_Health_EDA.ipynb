```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np

sns.set_theme(style="darkgrid")
plt.rcParams["figure.figsize"] = (10, 6)

print("✅ Analytics and visualization libraries successfully initialized!")
```

```python
import kagglehub
import os
# Download latest version
path = kagglehub.dataset_download("shivasingh4945/student-social-media-and-mental-health-impact")


csv_file_name = os.listdir(path)[0]
csv_path = os.path.join(path, csv_file_name)


df = pd.read_csv(csv_path)
```

```python
print(df.shape)
print(df.info())
print(df.head(3))
```

```python
platform_counts = df['Most_Used_Platform'].value_counts()

x_data = platform_counts.index
y_data = platform_counts.values

fig, ax = plt.subplots(figsize=(12, 6))

ax.bar(x_data, y_data, color="royalblue", edgecolor="white")

ax.set_title("Most Used Social Media Platforms")
ax.set_xlabel("Social Media Patform")
ax.set_ylabel("Usage")

ax.tick_params(axis='x', rotation=45)


plt.show()
```

```python
platform_counts = df['Most_Used_Platform'].value_counts()

x_data = platform_counts.index
y_data = platform_counts.values

fig, ax = plt.subplots(figsize=(12, 6))

ax.barh(x_data, y_data, color="royalblue", edgecolor="white")
ax.invert_yaxis()

ax.set_title("Most Used Social Media Platforms")
ax.set_ylabel("Social Media Patform")
ax.set_xlabel("Usage")


plt.show()
```

```python
# Seaborn Implementation of Bar graph
fig, ax = plt.subplots(figsize=(10, 6))

sns.countplot(data=df, x="Most_Used_Platform", ax=ax, color='royalblue')

ax.set_title("Most Used Social Media Platforms (Seaborn)")
ax.set_xlabel("Social Media Platform")
ax.set_ylabel("Count")
ax.tick_params(axis='x', rotation=45)

plt.show()
```

```python
sleep_data = df['Sleep_Hours_Per_Night']

fig, ax = plt.subplots(figsize=(10, 6))

ax.hist(sleep_data, bins=25, edgecolor="white", color="royalblue")

"""
Histogram chart is a graphical representation of the distribution of numerical data, where data is grouped into bins or intervals, and each bin's height represents the frequency of data points within that range
"""


ax.set_title("Distribution of Student Sleep Hours")
ax.set_xlabel("Hours of Sleep Per Night")
ax.set_ylabel("Number of Students")

plt.show()
```

```python
# Seaborn Implementation

fig, ax = plt.subplots(figsize=(10, 6))

sns.histplot(data = df, x="Sleep_Hours_Per_Night", bins=25, edgecolor="white", ax=ax, color="royalblue")

ax.set_title("Distribution of Student Sleep Hours (Seaborn)")
ax.set_xlabel("Hours of Sleep Per Night")
ax.set_ylabel("Number of Students")

plt.show()
```

```python
x_data = df['Study_Hours']
y_data = df['Sleep_Hours_Per_Night']

fig, ax = plt.subplots(figsize=(10, 6))

ax.scatter(x_data, y_data, alpha=0.5, color="royalblue")

"""
Unlike a bar chart or histogram, ax.scatter() doesn't aggregate or count anything. It plots every single individual row as a lone dot (x, y) on a Cartesian plane.
"""

ax.set_title("Study Hours vs. Sleep Hours Per Night")
ax.set_xlabel("Daily Study Hours")
ax.set_ylabel("Sleep Hours Per Night")

plt.show()
```

```python
x_data = df['Avg_Daily_Usage_Hours']
y_data = df['Sleep_Hours_Per_Night']

fig, ax = plt.subplots(figsize=(10, 6))

ax.scatter(x_data, y_data, alpha=0.5, color="royalblue")

ax.set_title("Screen Time vs. Sleep Hours Per Night")
ax.set_xlabel("Screen Time")
ax.set_ylabel("Sleep Hours Per Night")

plt.show()
```

```python
# Seaborn Implementation
fig, ax = plt.subplots(figsize=(10, 6))

sns.scatterplot(data=df, x="Avg_Daily_Usage_Hours", y="Sleep_Hours_Per_Night", alpha=0.5, ax=ax, color="royalblue")

ax.set_title("Screen Time vs. Sleep Hours Per Night (Seaborn)")
ax.set_xlabel("Screen Time (Hours)")
ax.set_ylabel("Sleep Hours Per Night")

plt.show()
```

```python
low_stress_sleep = df[df['Stress_Level'] == 'Low']['Sleep_Hours_Per_Night']
med_stress_sleep = df[df['Stress_Level'] == 'Medium']['Sleep_Hours_Per_Night']
high_stress_sleep = df[df['Stress_Level'] == 'High']['Sleep_Hours_Per_Night']

data_to_plot = [low_stress_sleep, med_stress_sleep, high_stress_sleep]
labels = ["Low", "Medium", "High"]

fig, ax = plt.subplots(figsize=(10, 6))

ax.boxplot(data_to_plot, labels=labels)

"""
The Median (The Line inside the Box): This is the exact middle value of the data (50th percentile).

The Box (IQR): The box represents the middle 50% of your data. The bottom of the box is the 25th percentile, and the top is the 75th percentile.

The Whiskers (The Tines): These extend out to show the minimum and maximum range of the rest of the data (excluding extreme anomalies).

The Fliers/Outliers (Individual Dots): Any data points that sit way outside the normal range are plotted as individual dots past the whiskers.
"""

ax.set_title("Sleep Hours Across Different Stress Levels")
ax.set_xlabel("Stress Level")
ax.set_ylabel("Sleep Hours Per Night")

plt.show()
```

```python
# Seaborn Implementation

fig, ax = plt.subplots(figsize=(10, 6))

sns.boxplot(data=df, x="Stress_Level", y="Sleep_Hours_Per_Night", ax=ax, order=['Low', 'Medium', 'High'])

ax.set_title("Sleep Hours Across Different Stress Levels (Seaborn)")
ax.set_xlabel("Stress Level")
ax.set_ylabel("Sleep Hours Per Night")

plt.show()
```

```python
age_study_trends = df.groupby('Age')['Study_Hours'].mean()

x_data = age_study_trends.index
y_data = age_study_trends.values

fig, ax = plt.subplots(figsize=(10, 6))

ax.plot(x_data, y_data, marker='o')

"""
The X-Axis (The Progression): This must be sorted in order (like Age: 18, 19, 20, 21...) so the line moves naturally from left to right.

The Y-Axis (The Metric): This is the value we are tracking at each age (like average study hours).

The Line: Instead of individual floating dots, the line emphasizes the direction and trend (is it rising, falling, or staying flat?).
"""
ax.set_title("Average Daily Study Hours by Age")
ax.set_xlabel("Age of Student")
ax.set_ylabel("Average Study Hours")

plt.show()
```

```python
fig, ax = plt.subplots(figsize=(10, 5))

sns.lineplot(data=df, x="Age", y="Study_Hours", ax=ax, marker="o")

ax.set_title("Average Daily Study Hours by Age (Seaborn)")
ax.set_xlabel("Age of Student")
ax.set_ylabel("Average Study Hours")
ax.set_xticks(df["Age"].unique())

plt.show()
```

```python
grouped_data = df.groupby(["Age", "Gender"])['Study_Hours'].mean()

ages = sorted(df['Age'].unique())

female_trends = grouped_data.xs('Female', level="Gender")
male_trends = grouped_data.xs('Male', level="Gender")

fig, ax = plt.subplots(figsize=(10, 6))

ax.plot(ages, female_trends, marker='o', label="Female")
ax.plot(ages, male_trends, marker='o', label="Male")

ax.set_title("Average Daily Study Hours by Age and Gender")
ax.set_xlabel("Age of Student")
ax.set_ylabel("Average Study Hours")

ax.legend()
plt.show()
```

```python
# Seaborn Implementation

fig, ax = plt.subplots(figsize=(10, 6))

sns.lineplot(data=df, x="Age", y="Study_Hours", hue="Gender", ax=ax, marker="o")

ax.set_title("Study Trends by Age and Gender (Seaborn)")
ax.set_xlabel("Age")
ax.set_ylabel("Average Daily Study Hours")

plt.show()
```

```python
numeric_df = df.select_dtypes(include=['float64', 'int64'])
corr_matrix = numeric_df.corr()

fig, ax = plt.subplots(figsize=(10, 8))

sns.heatmap(corr_matrix, annot=True, cmap='coolwarm', fmt=".2f", ax=ax)

ax.set_title("Correlation Heatmap of Student Metrics", fontsize=16)

plt.show()
```
